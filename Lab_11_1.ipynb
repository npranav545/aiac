```python
class Stack:
    """A simple Stack implementation using a Python list."""

    def __init__(self):
        """Initializes an empty stack."""
        self._items = []

    def push(self, item):
        """
        Adds an item to the top of the stack.

        Args:
            item: The item to be added.
        """
        self._items.append(item)

    def pop(self):
        """
        Removes and returns the item from the top of the stack.

        Returns:
            The item removed from the stack.

        Raises:
            IndexError: If the stack is empty.
        """
        if not self.is_empty():
            return self._items.pop()
        else:
            raise IndexError("pop from empty stack")

    def peek(self):
        """
        Returns the item at the top of the stack without removing it.

        Returns:
            The item at the top of the stack.

        Raises:
            IndexError: If the stack is empty.
        """
        if not self.is_empty():
            return self._items[-1]
        else:
            raise IndexError("peek from empty stack")

    def is_empty(self):
        """
        Checks if the stack is empty.

        Returns:
            True if the stack is empty, False otherwise.
        """
        return len(self._items) == 0
```

```python
# Test Case 1: Basic push and pop
stack = Stack()
stack.push(10)
stack.push(20)
assert stack.pop() == 20
assert stack.pop() == 10
assert stack.is_empty()
print("Test Case 1 Passed: Basic push and pop works.")

# Test Case 2: Peek operation
stack = Stack()
stack.push(30)
assert stack.peek() == 30
assert not stack.is_empty()
print("Test Case 2 Passed: Peek operation works.")

# Test Case 3: Popping from empty stack (should raise an exception)
stack = Stack()
try:
    stack.pop()
    assert False, "Test Case 3 Failed: Pop from empty stack should raise an exception."
except IndexError:
    assert True
    print("Test Case 3 Passed: Popping from empty stack raises an exception.")
```

```python
class Queue:
    """A simple Queue implementation using a Python list."""

    def __init__(self):
        """Initializes an empty queue."""
        self._items = []

    def enqueue(self, item):
        """
        Adds an item to the rear of the queue.

        Args:
            item: The item to be added.
        """
        self._items.append(item)

    def dequeue(self):
        """
        Removes and returns the item from the front of the queue (FIFO).

        Returns:
            The item removed from the queue.

        Raises:
            IndexError: If the queue is empty.
        """
        if not self.is_empty():
            return self._items.pop(0)  # Remove from the beginning for FIFO
        else:
            raise IndexError("dequeue from empty queue")

    def peek(self):
        """
        Returns the item at the front of the queue without removing it.

        Returns:
            The item at the front of the queue.

        Raises:
            IndexError: If the queue is empty.
        """
        if not self.is_empty():
            return self._items[0]
        else:
            raise IndexError("peek from empty queue")

    def is_empty(self):
        """
        Checks if the queue is empty.

        Returns:
            True if the queue is empty, False otherwise.
        """
        return len(self._items) == 0

    def size(self):
        """
        Returns the number of items in the queue.

        Returns:
            The number of items in the queue.
        """
        return len(self._items)
```

```python
# Test Case 1: Basic enqueue and dequeue
queue = Queue()
queue.enqueue(100)
queue.enqueue(200)
assert queue.dequeue() == 100
assert queue.dequeue() == 200
assert queue.is_empty()
print("Test Case 1 Passed: Basic enqueue and dequeue works.")

# Test Case 2: Peek and size operations
queue = Queue()
queue.enqueue(300)
queue.enqueue(400)
assert queue.peek() == 300
assert queue.size() == 2
assert queue.dequeue() == 300
assert queue.size() == 1
print("Test Case 2 Passed: Peek and size operations work.")

# Test Case 3: Dequeue from empty queue (should raise an exception)
queue = Queue()
try:
    queue.dequeue()
    assert False, "Test Case 3 Failed: Dequeue from empty queue should raise an exception."
except IndexError:
    assert True
    print("Test Case 3 Passed: Dequeuing from empty queue raises an exception.")
```

```python
class Node:
    """Represents a node in a singly linked list."""

    def __init__(self, data=None):
        """
        Initializes a new Node.

        Args:
            data: The data to be stored in the node (default is None).
        """
        self.data = data
        self.next = None  # Pointer to the next node

class LinkedList:
    """Represents a singly linked list."""

    def __init__(self):
        """Initializes an empty linked list."""
        self.head = None  # The head of the list

    def insert_at_end(self, data):
        """
        Inserts a new node with the given data at the end of the list.

        Args:
            data: The data to be inserted.
        """
        new_node = Node(data)
        if self.head is None:
            self.head = new_node
            return

        last_node = self.head
        while last_node.next:
            last_node = last_node.next
        last_node.next = new_node

    def display(self):
        """
        Displays the elements of the linked list.

        Returns:
            A string representation of the linked list elements.
        """
        elements = []
        current = self.head
        while current:
            elements.append(str(current.data))
            current = current.next
        return " -> ".join(elements)
```

```python
# Test Case 1: Insert a single node and display
llist1 = LinkedList()
llist1.insert_at_end(10)
assert llist1.display() == "10"
print("Test Case 1 Passed: Single node insertion and display works.")

# Test Case 2: Insert multiple nodes and display
llist2 = LinkedList()
llist2.insert_at_end(20)
llist2.insert_at_end(30)
llist2.insert_at_end(40)
assert llist2.display() == "20 -> 30 -> 40"
print("Test Case 2 Passed: Multiple node insertion and display works.")

# Test Case 3: Display an empty list
llist3 = LinkedList()
assert llist3.display() == ""
print("Test Case 3 Passed: Displaying an empty list works.")
```

```python
class Node:
    """Represents a node in a Binary Search Tree."""

    def __init__(self, data):
        """
        Initializes a new Node.

        Args:
            data: The data to be stored in the node.
        """
        self.data = data
        self.left = None   # Pointer to the left child
        self.right = None  # Pointer to the right child

class BST:
    """Represents a Binary Search Tree."""

    def __init__(self):
        """Initializes an empty BST."""
        self.root = None  # The root of the tree

    def insert(self, data):
        """
        Recursively inserts a new node with the given data into the BST.

        Args:
            data: The data to be inserted.
        """
        if self.root is None:
            self.root = Node(data)
        else:
            self._insert_recursive(self.root, data)

    def _insert_recursive(self, current_node, data):
        """
        Helper method for recursive insertion.

        Args:
            current_node: The current node being considered.
            data: The data to be inserted.
        """
        if data < current_node.data:
            if current_node.left is None:
                current_node.left = Node(data)
            else:
                self._insert_recursive(current_node.left, data)
        elif data > current_node.data:
            if current_node.right is None:
                current_node.right = Node(data)
            else:
                self._insert_recursive(current_node.right, data)
        # If data is equal, do nothing (assuming no duplicates allowed)

    def inorder_traversal(self):
        """
        Performs an in-order traversal of the BST and returns a list of elements.

        Returns:
            A list containing the elements of the BST in sorted order.
        """
        elements = []
        self._inorder_recursive(self.root, elements)
        return elements

    def _inorder_recursive(self, current_node, elements):
        """
        Helper method for recursive in-order traversal.

        Args:
            current_node: The current node being considered.
            elements: The list to store the elements.
        """
        if current_node:
            self._inorder_recursive(current_node.left, elements)
            elements.append(current_node.data)
            self._inorder_recursive(current_node.right, elements)
```

```python
class HashTable:
    """A simple HashTable implementation with chaining for collision handling."""

    def __init__(self, size):
        """
        Initializes a new HashTable.

        Args:
            size: The size of the hash table (number of buckets).
        """
        self.size = size
        self._table = [[] for _ in range(self.size)]  # Initialize with empty lists for chaining

    def _hash(self, key):
        """
        Calculates the hash value for a given key.

        Args:
            key: The key to be hashed.

        Returns:
            The hash value (index of the bucket).
        """
        return hash(key) % self.size

    def insert(self, key, value):
        """
        Inserts a key-value pair into the hash table. Handles collisions using chaining.

        Args:
            key: The key to be inserted.
            value: The value associated with the key.
        """
        index = self._hash(key)
        for i, (existing_key, existing_value) in enumerate(self._table[index]):
            if existing_key == key:
                self._table[index][i] = (key, value)  # Update value if key exists
                return
        self._table[index].append((key, value)) # Add new key-value pair

    def search(self, key):
        """
        Searches for a key in the hash table.

        Args:
            key: The key to search for.

        Returns:
            The value associated with the key, or None if the key is not found.
        """
        index = self._hash(key)
        for existing_key, existing_value in self._table[index]:
            if existing_key == key:
                return existing_value
        return None  # Key not found

    def delete(self, key):
        """
        Deletes a key-value pair from the hash table.

        Args:
            key: The key to delete.

        Returns:
            The value associated with the deleted key, or None if the key was not found.
        """
        index = self._hash(key)
        for i, (existing_key, existing_value) in enumerate(self._table[index]):
            if existing_key == key:
                del self._table[index][i]
                return existing_value
        return None # Key not found
```

```python
class Graph:
    """Represents a graph using an adjacency list."""

    def __init__(self):
        """Initializes an empty graph."""
        self._graph = {}  # Adjacency list: {vertex: [neighbors]}

    def add_vertex(self, vertex):
        """
        Adds a vertex to the graph if it doesn't already exist.

        Args:
            vertex: The vertex to be added.
        """
        if vertex not in self._graph:
            self._graph[vertex] = []

    def add_edge(self, vertex1, vertex2):
        """
        Adds an undirected edge between two vertices. Adds vertices if they don't exist.

        Args:
            vertex1: The first vertex of the edge.
            vertex2: The second vertex of the edge.
        """
        self.add_vertex(vertex1)
        self.add_vertex(vertex2)
        if vertex2 not in self._graph[vertex1]:
            self._graph[vertex1].append(vertex2)
        if vertex1 not in self._graph[vertex2]:
            self._graph[vertex2].append(vertex1)

    def display_connections(self):
        """
        Displays the connections (adjacency list) of the graph.

        Returns:
            A string representation of the graph's adjacency list.
        """
        output = []
        for vertex, neighbors in self._graph.items():
            output.append(f"{vertex}: {neighbors}")
        return "\n".join(output)

    def get_vertices(self):
        """
        Returns a list of all vertices in the graph.
        """
        return list(self._graph.keys())

    def get_neighbors(self, vertex):
        """
        Returns a list of neighbors for a given vertex.

        Args:
            vertex: The vertex to get neighbors for.

        Returns:
            A list of neighbors, or None if the vertex doesn't exist.
        """
        return self._graph.get(vertex)
```

```python
import heapq

class PriorityQueue:
    """A Priority Queue implementation using Python's heapq module."""

    def __init__(self):
        """Initializes an empty priority queue."""
        self._queue = []  # The list to store elements, maintained as a heap

    def enqueue(self, priority, item):
        """
        Adds an item to the priority queue with a given priority.

        Args:
            priority: The priority of the item (lower number means higher priority).
            item: The item to be added.
        """
        # heapq works with tuples (priority, item). Lower priority numbers are
        # considered 'smaller' and thus have higher priority in a min-heap.
        heapq.heappush(self._queue, (priority, item))

    def dequeue(self):
        """
        Removes and returns the item with the highest priority from the queue.

        Returns:
            The item with the highest priority.

        Raises:
            IndexError: If the priority queue is empty.
        """
        if not self.is_empty():
            return heapq.heappop(self._queue)[1]  # Return the item (index 1 of the tuple)
        else:
            raise IndexError("dequeue from empty priority queue")

    def is_empty(self):
        """
        Checks if the priority queue is empty.

        Returns:
            True if the priority queue is empty, False otherwise.
        """
        return len(self._queue) == 0

    def display(self):
        """
        Displays the elements currently in the priority queue (in their heap representation).
        Note: This does not display elements in priority order, as heapq maintains
        a heap property, not strict sorting.

        Returns:
            A list representing the internal state of the priority queue.
        """
        return list(self._queue)
```

```python
from collections import deque

class DequeDS:
    """A Deque (Double-Ended Queue) implementation using collections.deque."""

    def __init__(self):
        """Initializes an empty deque."""
        self._deque = deque()

    def insert_front(self, item):
        """
        Adds an item to the front of the deque.

        Args:
            item: The item to be added.
        """
        self._deque.appendleft(item)

    def insert_rear(self, item):
        """
        Adds an item to the rear of the deque.

        Args:
            item: The item to be added.
        """
        self._deque.append(item)

    def remove_front(self):
        """
        Removes and returns the item from the front of the deque.

        Returns:
            The item removed from the deque.

        Raises:
            IndexError: If the deque is empty.
        """
        if not self.is_empty():
            return self._deque.popleft()
        else:
            raise IndexError("remove from empty deque")

    def remove_rear(self):
        """
        Removes and returns the item from the rear of the deque.

        Returns:
            The item removed from the deque.

        Raises:
            IndexError: If the deque is empty.
        """
        if not self.is_empty():
            return self._deque.pop()
        else:
            raise IndexError("remove from empty deque")

    def is_empty(self):
        """
        Checks if the deque is empty.

        Returns:
            True if the deque is empty, False otherwise.
        """
        return len(self._deque) == 0

    def display(self):
        """
        Displays the elements of the deque.

        Returns:
            A list representation of the deque's elements.
        """
        return list(self._deque)
```

```python
# Test Case 1: Basic insertions and removals from both ends
deque1 = DequeDS()
deque1.insert_front(10)
deque1.insert_rear(20)
deque1.insert_front(5)
deque1.insert_rear(25)
assert deque1.display() == [5, 10, 20, 25]
assert deque1.remove_front() == 5
assert deque1.remove_rear() == 25
assert deque1.display() == [10, 20]
print("Test Case 1 Passed: Basic insertions and removals from both ends work.")

# Test Case 2: Removing from an empty deque (should raise an exception)
deque2 = DequeDS()
try:
    deque2.remove_front()
    assert False, "Test Case 2 Failed: Removing from empty deque should raise an exception."
except IndexError:
    assert True
    print("Test Case 2 Passed: Removing from empty deque raises an exception.")

deque3 = DequeDS()
try:
    deque3.remove_rear()
    assert False, "Test Case 2 Failed: Removing from empty deque should raise an exception."
except IndexError:
    assert True
    print("Test Case 2 Passed: Removing from empty deque raises an exception.")


# Test Case 3: Insert and remove until empty
deque4 = DequeDS()
deque4.insert_front(1)
deque4.insert_rear(2)
deque4.remove_front()
deque4.remove_rear()
assert deque4.is_empty()
assert deque4.display() == []
print("Test Case 3 Passed: Inserting and removing until empty works.")
```

| Feature                     | Most Appropriate Data Structure | Justification                                                                                                                               |
|-----------------------------|---------------------------------|---------------------------------------------------------------------------------------------------------------------------------------------|
| Student Attendance Tracking | Hash Table (Dictionary)         | Hash tables provide efficient O(1) average time complexity for inserting and searching for student attendance records using student ID as the key. It allows quick lookup and update of attendance status. |
| Event Registration System   | Queue                           | A queue is suitable for managing event registrations in a first-come, first-served (FIFO) manner. As registrations come in, they are added to the rear, and processed from the front. |
| Library Book Borrowing      | Hash Table (or Linked List with Search) | A Hash Table (dictionary) is efficient for searching for books by ISBN or title (O(1) average). A Linked List could store borrowed books per user, allowing easy addition/removal (O(1) at head/end), but searching would be O(n). |
| Bus Scheduling System       | Graph                           | A graph is ideal for representing bus routes and stops. Vertices can be stops, and edges can be routes with associated weights (distance, time). Graph algorithms can be used for pathfinding and scheduling. |
| Cafeteria Order Queue       | Queue                           | A queue is the most appropriate structure for managing customer orders in a cafeteria, ensuring orders are processed in the order they were placed (FIFO).                                    |

| Data Structure   | Insert (Avg/Worst) | Delete (Avg/Worst) | Search (Avg/Worst) | Traversal |
|------------------|--------------------|--------------------|--------------------|-----------|
| **Stack**        | O(1)/O(1)          | O(1)/O(1)          | N/A                | O(n) (if iterating) |
| **Queue**        | O(1)/O(1)          | O(1)/O(1)          | N/A                | O(n) (if iterating) |
| **Linked List**  | O(1)/O(n) (end/arbitrary) | O(1)/O(n) (head/arbitrary) | O(n)/O(n)          | O(n)      |
| **BST**          | O(log n)/O(n)      | O(log n)/O(n)      | O(log n)/O(n)      | O(n)      |
| **Hash Table**   | O(1)/O(n)          | O(1)/O(n)          | O(1)/O(n)          | O(n) (if iterating all buckets) |
| **Graph**        | O(V+E) (add vertex/edge) | O(V+E) (remove vertex/edge) | O(V+E) (BFS/DFS) | O(V+E) (BFS/DFS) |
| **Priority Queue**| O(log n)/O(log n)  | O(log n)/O(log n)  | N/A                | O(n log n) (if sorting) |
| **Deque**        | O(1)/O(1)          | O(1)/O(1)          | N/A                | O(n) (if iterating) |

**Notes:**
*   Time complexities are generally expressed in Big O notation.
*   `n` refers to the number of elements in the data structure.
*   `V` refers to the number of vertices and `E` refers to the number of edges in a graph.
*   "Avg" refers to the average case time complexity, and "Worst" refers to the worst-case time complexity.
*   N/A means the operation is not a primary or typical operation for the data structure in that context.
*   Traversal for linear structures like Stack, Queue, Linked List, and Deque refers to iterating through all elements.
*   Traversal for Graph is typically done using BFS or DFS.
*   Traversal for Priority Queue in order of priority would involve repeatedly dequeuing, leading to a higher complexity. Displaying the internal representation (heap) is faster but not in sorted order.

```python
"""This module contains functions for calculating sums of even and odd numbers and a class for managing student details."""

def sum_even_odd(numbers):
  """Calculates the sum of even and odd numbers in a list.

  This function iterates through a list of numbers and separates them into even and odd.
  It then calculates the sum of the even numbers and the sum of the odd numbers.

  Args:
    numbers: A list of integers to process.

  Returns:
    A tuple containing two integers:
    - The first element is the sum of the even numbers.
    - The second element is the sum of the odd numbers.
  """
  even_sum = 0
  odd_sum = 0
  for number in numbers:
    if number % 2 == 0:
      even_sum += number
    else:
      odd_sum += number
  return even_sum, odd_sum

user_input = input("Enter a list of numbers separated by spaces: ")
numbers_str = user_input.split()
user_numbers = [int(num) for num in numbers_str]

even_sum, odd_sum = sum_even_odd(user_numbers)

print(f"Sum of even numbers: {even_sum}")
print(f"Sum of odd numbers: {odd_sum}")
```

```python
class sru_student:
    """Represents a student at SRU with details like name, roll number, hostel status, and fees.""" # Class docstring explaining the purpose of the class

    def __init__(self, name, roll_number, hostel_status, fees):
        """Initializes a new sru_student object.

        Args:
            name: The name of the student.
            roll_number: The roll number of the student.
            hostel_status: The hostel status of the student (e.g., "Day Scholar", "Hostelite").
            fees: The initial fees paid by the student.
        """ # Docstring for the constructor
        self.name = name  # Assigning the name to the instance
        self.roll_number = roll_number  # Assigning the roll number to the instance
        self.hostel_status = hostel_status  # Assigning the hostel status to the instance
        self.fees = fees  # Assigning the initial fees to the instance

    def fee_update(self, amount):
        """Updates the fees paid by the student.

        Args:
            amount: The amount to add to the current fees.
        """ # Docstring for the fee_update method
        self.fees += amount  # Adding the update amount to the current fees
        print(f"Fees updated for {self.name}. New total fees: {self.fees}")  # Printing the updated fee information

    def display_details(self):
        """Displays the details of the student.""" # Docstring for the display_details method
        print("Student Details:")  # Printing a header for student details
        print(f"Name: {self.name}")  # Printing the student's name
        print(f"Roll Number: {self.roll_number}")  # Printing the student's roll number
        print(f"Hostel Status: {self.hostel_status}")  # Printing the student's hostel status
        print(f"Fees Paid: {self.fees}")  # Printing the student's fees paid
```

Now, let's create an instance of the `sru_student` class and use its methods.

```python
# Create a student object
student1 = sru_student("Alice", "SRU123", "Day Scholar", 50000) # Creating an instance of the sru_student class

# Display initial details
student1.display_details() # Calling the display_details method to show initial information

# Update fees
student1.fee_update(20000) # Calling the fee_update method to add to the fees

# Display details after fee update
student1.display_details() # Calling the display_details method again to show updated information
```

```python
def add(x, y):
  """
  Adds two numbers.

  Parameters
  ----------
  x : float
      The first number.
  y : float
      The second number.

  Returns
  -------
  float
      The sum of the two numbers.
  """
  return x + y

def subtract(x, y):
  """
  Subtracts the second number from the first.

  Parameters
  ----------
  x : float
      The first number.
  y : float
      The second number.

  Returns
  -------
  float
      The difference between the two numbers.
  """
  return x - y

def multiply(x, y):
  """
  Multiplies two numbers.

  Parameters
  ----------
  x : float
      The first number.
  y : float
      The second number.

  Returns
  -------
  float
      The product of the two numbers.
  """
  return x * y

def divide(x, y):
  """
  Divides the first number by the second.

  Parameters
  ----------
  x : float
      The first number (dividend).
  y : float
      The second number (divisor).

  Returns
  -------
  float or str
      The result of the division, or an error message if the divisor is zero.
  """
  if y == 0:
    return "Error: Division by zero is not allowed."
  return x / y

# Take input from the user
try:
  num1 = float(input("Enter the first number: "))
  num2 = float(input("Enter the second number: "))

  print(f"{num1} + {num2} = {add(num1, num2)}")
  print(f"{num1} - {num2} = {subtract(num1, num2)}")
  print(f"{num1} * {num2} = {multiply(num1, num2)}")
  print(f"{num1} / {num2} = {divide(num1, num2)}")
except ValueError:
  print("Invalid input. Please enter numbers.")
```

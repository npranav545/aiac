```python
def merge_sort(arr):
    """Sorts a list using the Merge Sort algorithm.
    Args:
        arr: The list to be sorted.
    Returns:
        The sorted list.
    Time Complexity: O(n log n) The algorithm divides the list in half repeatedly (log n divisions) and merging takes linear time (n).
    Space Complexity: O(n) - Additional space is required to store the temporary subarrays during the merge process.
    """
    if len(arr) <= 1:
        return arr
    mid = len(arr) // 2
    left_half = arr[:mid]
    right_half = arr[mid:]
    left_half = merge_sort(left_half)
    right_half = merge_sort(right_half)
    return merge(left_half, right_half)

def merge(left, right):
    """Merges two sorted lists into a single sorted list."""
    result = []
    i = j = 0
    while i < len(left) and j < len(right):
        if left[i] < right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1
    result.extend(left[i:])
    result.extend(right[j:])
    return result

# Get input from the user
input_str = input("Enter a list of numbers separated by spaces: ")
try:
    numbers = [int(x) for x in input_str.split()]
    sorted_numbers = merge_sort(numbers)
    print("Sorted list: ", sorted_numbers)
except ValueError:
    print("Invalid input. Please enter numbers separated by spaces.")
```

```python
def binary_search(arr, target):
    """Finds the index of a target number in a sorted list using Binary Search.
    Args:
        arr: The sorted list to search within.
        target: The number to search for.
    Returns:
        The index of the target if found, otherwise -1.
    Time Complexity:
        Best Case: O(1) - The target is found at the middle of the list in the first comparison.
        Average Case: O(log n)
        Worst Case: O(log n) - The list is repeatedly halved until the target is found or the search space is empty. The target is at either end of the list or not present, requiring the maximum number of comparisons.
    """
    low = 0
    high = len(arr) - 1
    while low <= high:
        mid = (low + high) // 2
        if arr[mid] == target:
            return mid
        elif arr[mid] < target:
            low = mid + 1
        else:
            high = mid - 1
    return -1

# Get input from the user for the sorted list
input_list_str = input("Enter a sorted list of numbers separated by spaces: ")
try:
    sorted_numbers = [int(x) for x in input_list_str.split()]
    # Get input from the user for the target number
    target_str = input("Enter the target number to search for: ")
    target = int(target_str)

    # Perform binary search
    index = binary_search(sorted_numbers, target)

    # Display the result
    if index != -1:
        print(f"Target {target} found at index: {index}")
    else:
        print(f"Target {target} not found in the list.")

except ValueError:
    print("Invalid input. Please enter numbers separated by spaces for the list and a number for the target.")
```

```python
def sort_products_by_price(products):
    """Sorts a list of products by price.

    For large datasets, efficient sorting algorithms like Merge Sort or Timsort
    (used by Python's built-in sorted() function) are recommended due to their
    average and worst-case time complexity of O(n log n). This is significantly
    better than algorithms like Bubble Sort or Insertion Sort which have O(n^2)
    time complexity in the worst case, making them impractical for large n.

    Args:
        products: A list of dictionaries, where each dictionary represents a product.

    Returns:
        A new list of products sorted by price.
    """
    # Using sorted() which implements Timsort, an efficient algorithm for large datasets.
    return sorted(products, key=lambda x: x['Price'])

def sort_products_by_id(products):
    """Sorts a list of products by ID.

    Similar to sorting by price, for large datasets, efficient sorting
    algorithms like Merge Sort or Timsort (used by Python's built-in sorted()
    function) are recommended due to their average and worst-case time
    complexity of O(n log n). These algorithms are much faster for large n
    compared to O(n^2) algorithms.

    Args:
        products: A list of dictionaries, where each dictionary represents a product.

    Returns:
        A new list of products sorted by ID.
    """
    # Using sorted() which implements Timsort, an efficient algorithm for large datasets.
    return sorted(products, key=lambda x: x['ID'])

# Example usage (optional - for testing)
# sorted_by_price = sort_products_by_price(products)
# print("Sorted by Price:", sorted_by_price)
# sorted_by_id = sort_products_by_id(products)
# print("Sorted by ID:", sorted_by_id)
```
